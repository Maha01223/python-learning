File handling is a crucial part of any pragramming language.
python provides buil- in funcion and methods to read from and wrie to files,both text and binary.
this lesson will cover the basis of file handling induling reading and writing text abd binary files.

```python
### Read a whole file
with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
#Read a Whole file
with open('example.txt', 'r') as file:
    for line in file:
        print(line)
        print(line.strip()) #strip() remove the newlines charatcter
```

```python
#writing a file (overwriting)
with open('example.txt', 'w') as file:
    file.write('hello world!\n')
    file.write('This is a newline\n')
```

```python
#write a file without overwriting
with open('example.txt', 'a') as file:
    file.write("Append opration taken place!\n")
```

```python
#write a list of line to a file
lines = ['First line\n', 'second line \n', 'third line \n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
#binary files
#first create a example.bin file

#writing to a binary files

data = b'\x00\x01\x02\x03\x04\x05'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
#Reading a binary file

with open('example.bin', 'rb') as file:
     content = file.read()
     print(content)
```

```python
#read the content from a source.txt file and write to a destination text file
#Copying a text file
with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```
