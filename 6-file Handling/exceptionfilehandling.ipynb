Exception handling in python allow you to handle error gracefully and take corrective action, wihou stopping the execution of the program.

Excepion are events hat disrupt the normal flow of a program
1. ZeroDivisionerror

```python
a=b
print(a)
```

```python
#Exception try, Exception Block

try:
    a=b
except:
    print("the variable has not been assigned")
```

```python
try:
    a=b
except NameError as ex:
    print(ex)
```
