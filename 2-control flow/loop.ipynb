```python
#While Loop
#the while loop continous to execute as log as the  condition is True

count = 0
while count<=10:
    print(count)
    count = count + 2
```

```python
#loop control Statement
#the break statement exist the loop permanenty
#break Statement

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
#Continous
#the continous Satement skip the current iteration and continous with the next

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
#pass 
#the pa statement is a null operation.

for i in range(5):
    if i == 3:
        pass
    print(i)
```

```python
#Nested loop
#The loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i : {i} and j : {j}")
```

```python
# calculate the sum of first n natura number using a while and for loop
#while loop

n = 10
sum = 0
count = 1

while count<=n:
    sum = sum + count
    count = count + 1
    print("sum of first 10 natura number:", sum)
```

```python
#Nested If statement

num = int(input("Enter the value"))

if (num < 0):
    print("negative")
elif (num > 0):
    if (num <= 10):
        print("number greater and equl 1-10")
    elif (num > 10 and num <= 20):
        print("number is between 11-20")
    else:
        print("number is greater then 20")
else:
    print("number is zero")
```
