In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
DF_filled = df.fillna(0)
print(DF_filled)
```

```python
df['sales_filna']=df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

```python
##reanamin column
df = df.rename(columns={'Date':'sales_Date'})
df.head()
```

```python
#change datatypes
df['value_new']=df['Value'].fillna(df['Value'].mean()).astype(int)
df.head()
```

```python
#aply Lamda funcion

df['New Value']=df['Value'].apply(lambda x:x*2)
df.head()
```

```python
#data aggregating and grouping

df.head()
```

```python
group_mean = df.groupby('Product')['Value'].mean()
print(group_mean)
```

```python
#grouping
group_sum = df.groupby(['Product', 'Region'])['Value'].sum()
print(group_sum)
```

```python
df.groupby(['Product', 'Region'])['Value'].mean()
```
